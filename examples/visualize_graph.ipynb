# Computational Graph Visualization in Gradience

This notebook demonstrates how to use the visualization subsystem in **Gradience** to extract and render the dynamic computation graph of any tensor operation. We will construct a simple linear classification forward pass, a highly complex mathematical function with branching dependencies, and show how to render and export the results.

## 1. Environment Setup

```python
import sys
import os
sys.path.append(os.path.abspath('..'))

import numpy as np
from gradience import Tensor
```

## 2. Example 1: Simple Linear Layer with Activation

We instantiate a small input tensor, weight parameter, and bias. We then perform a matrix multiplication, add the bias, and pass it through a sigmoid activation.

```python
x = Tensor([[1.5, -0.5]], requires_grad=True)
w = Tensor([[0.8], [0.2]], requires_grad=True)
b = Tensor([0.5], requires_grad=True)

z = x @ w + b
y_hat = z.sigmoid()
```

We call `.visualize()` on the final output tensor. Jupyter notebook environments render the returned Graphviz object inline automatically.

```python
y_hat.visualize()
```

## 3. Example 2: A Complex Mathematical Function with Branching and Reconverging Paths

Next, we construct a more complex mathematical expression:

$$f(x, y) = \sum \left( \frac{\exp(x \cdot y)}{\sqrt{x^2 + y^2}} \right)$$

This graph features:
- Element-wise operations (multiplication, addition, exponential, square root, division, and power).
- Multiple branches diverging from input parameters and reconverging at the division operator.
- A final sum reduction operation.

```python
x = Tensor([1.5, 2.0], requires_grad=True)
y = Tensor([0.5, -1.0], requires_grad=True)

prod = x * y
numerator = prod.exp()

x_sq = x ** 2
y_sq = y ** 2
denominator = (x_sq + y_sq).sqrt()

out = numerator / denominator
loss = out.sum()
```

We invoke the backward pass to calculate gradients, then visualize the resulting graph.

```python
loss.backward()
loss.visualize()
```

## 4. Exporting the Graph to a File

We can save the computation graph to an external file (e.g. PDF or PNG) by calling `.render()` on the Graphviz Digraph object.

```python
graph = loss.visualize()
graph.render("complex_computation_graph", format="png", cleanup=True)
```
